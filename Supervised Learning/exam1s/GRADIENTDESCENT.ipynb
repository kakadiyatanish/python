```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import PolynomialFeatures

from sklearn.linear_model import LinearRegression

from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.model_selection import learning_curve
```

```python
df = pd.read_csv("RealEstate_HousePrice_Dataset.csv")
```

```python
df.head()
```

```python
target = "house_price_inr"
```

```python
df_clean = df.copy()

print("\n")
print("="*60)
print("PART G : GRADIENT DESCENT")
print("="*60)

X = df_clean['area_sqft'].values
y = df_clean[target].values

X = (X-X.mean())/X.std()
y = (y-y.mean())/y.std()

m = len(X)

theta0 = 0
theta1 = 0

learning_rate = 0.01

epochs = 1000

cost_history = []

for epoch in range(epochs):

    prediction = theta0 + theta1*X

    d0 = (1/m)*np.sum(prediction-y)
    d1 = (1/m)*np.sum((prediction-y)*X)

    theta0 -= learning_rate*d0
    theta1 -= learning_rate*d1

    cost = (1/(2*m))*np.sum(
        (prediction-y)**2
    )

    cost_history.append(cost)

print("Theta0 :",theta0)
print("Theta1 :",theta1)

plt.figure(figsize=(8,5))
plt.plot(cost_history)
plt.title("Gradient Descent Cost Reduction")
plt.xlabel("Epoch")
plt.ylabel("Cost")
plt.show()
```
