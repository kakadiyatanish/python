Distribution Plot

```python
for col in num_cols:

    plt.figure(figsize=(6,4))

    sns.histplot(
        df[col],
        kde=True
    )

    plt.title(f"Distribution of {col}")
    plt.show()
```

Fill Missing Values


```python
for col in num_cols:

    df[col].fillna(df[col].median(), inplace=True)
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:

    df[col].fillna(df[col].mode()[0], inplace=True)
```

Remove Outliers Using IQR Method

```python

num_cols = df.select_dtypes(include=['int64', 'float64']).columns

num_cols = num_cols.drop('SalePrice')

print("Original Shape :", df.shape)

for col in num_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("Shape After Removing Outliers :", df.shape)
```

# Feature Engineering

Create House Age

```python
if 'YearBuilt' in df.columns:

    df['HouseAge'] = 2026 - df['YearBuilt']
```

Create Remodel Age

```python
if 'YearRemodAdd' in df.columns:

    df['RemodelAge'] = 2026 - df['YearRemodAdd']
```

Create Total Bathrooms

```python
if {'FullBath','HalfBath'}.issubset(df.columns):

    df['TotalBath'] = df['FullBath'] + (0.5*df['HalfBath'])
```

Create Total Porch Area

```python
porch_cols=['OpenPorchSF',
            'EnclosedPorch',
            '3SsnPorch',
            'ScreenPorch']

available=[c for c in porch_cols if c in df.columns]

if len(available)>0:

    df['TotalPorchSF']=df[available].sum(axis=1)
```
